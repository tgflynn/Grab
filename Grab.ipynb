```python
# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------

NB_NAME = 'GRAB'
SITE_NAME = "example"
SITE_URL = "https://example.com/"
DOWNLOAD_DIR = Path(f"/content/{NB_NAME}/{SITE_NAME}/downloads")

# File extensions that we want to download
FILE_EXTENSIONS = {
    ""
    ".html"
    ".htm"
    ".pdf",
    ".epub",
    ".zip",
    ".txt",
    ".csv",
    ".json",
    ".mp3",
    ".mp4",
}

DOWNLOAD_DIR.mkdir(parents=True, exist_ok=True)
```

```python
# ------------------------------------------------------------
# HTTP session
# ------------------------------------------------------------

session = requests.Session()

session.headers.update({
    "User-Agent": (
        "Mozilla/5.0 "
        "(compatible; ColabDownloader/1.0)"
    )
})
```

```python
# ------------------------------------------------------------
# Functions
# ------------------------------------------------------------
def is_http_url(url):
    scheme = urlparse(url).scheme
    return scheme in ("http", "https")

def urlpath_to_fname(urlpath):
  urlpath2 = urlpath.replace("/", "-")
  return urlpath2

def download_url(url):
    if not is_http_url(url):
        return
    response = session.get(
        url,
        #stream=True,
        timeout=30
    )
    response.raise_for_status()

    url_fname = urlpath_to_fname(urlparse(url).path)
    print(f"url_fname = {url_fname}")
    target_path = os.path.join(DOWNLOAD_DIR, url_fname)
    print(f"target_path = {target_path}")
    #data = response.content.decode("utf-8
    data = response.content
    with open(target_path, "wb") as f:
        f.write(data)
```

```python
# ------------------------------------------------------------
# Get web site
# ------------------------------------------------------------

download_url(SITE_URL)
```

```python
# ------------------------------------------------------------
# Get a web page
# ------------------------------------------------------------
response = session.get(
    SITE_URL,
    timeout=30
)

response.raise_for_status()

print("Status:", response.status_code)
print("Final URL:", response.url)

#print(f"response.text = ${response.text}")


# ------------------------------------------------------------
# Parse the page
# ------------------------------------------------------------

soup = BeautifulSoup(response.text, "html.parser")

#print(f"soup.prettify() = ${soup.prettify()}")

links = []

for a in soup.find_all("a", href=True):
    url = urljoin(response.url, a["href"])
    links.append(url)

print(f"links = {links}")
print(f"links = ", links)


# ------------------------------------------------------------
# Select likely files
# ------------------------------------------------------------

target_links = []

for url in links:
    path = urlparse(url).path
    suffix = Path(path).suffix.lower()
    print(f"suffix = ***{suffix}***")
    print(f"type(suffix) = ***{type(suffix)}***")

    if is_http_url(url):
      target_links.append(url)


print(f"Found {len(target_links)} targets:")
print(f"target links: \n {target_links}")

for link in target_links:
  download_url(link)
```
